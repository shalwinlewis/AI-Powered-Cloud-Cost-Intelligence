```python
import pandas as pd
```

```python
df = pd.read_csv("focus_sample_100000.csv")
```

```python
print(df.dtypes)
```

```python
aws_df = df[df["ProviderName"] == "AWS"].copy()
```

```python
print("Rows:", aws_df.shape[0])
print("Columns:", aws_df.shape[1])
```

```python
service_cost = (
    aws_df.groupby("ServiceName")["BilledCost"]
    .sum()
    .sort_values(ascending=False)
)

print(service_cost)
```
